Linear **Regression**

```python
class Linear_Regression():

# intitiating the hyperparameters(learning rate, no of iterations)
  def __init__(self, learning_rate, no_of_iterations):
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations


  # fitting the dataset to the linear regression model
  def fit(self, X, Y):
     # no of training examples/data points and input features
     self.m, self.n = X.shape  # no of rows and columns

     # intitating the weight and bias
     self.w = np.zeros(self.n)
     self.b = 0
     self.X = X
     self.Y = Y

    # implementing Gradient Descent
    # for each time it goes throuh the dataset it updates the weights
     for i in range(self.no_of_iterations):
      self.update_weights()



  def update_weights(self,):
    Y_prediction = self.predict(self.X)

    # calculating gradint descents
    dw = -(2 * (self.X.T).dot(self.Y - Y_prediction)) / self.m
    db = -2 * np.sum(self.Y - Y_prediction) / self.m

    # updating the weights
    self.w = self.w - self.learning_rate * dw
    self.b = self.b - self.learning_rate * db


  def predict(self, X ):
    return X.dot(self.w) + self.b


```

Using linear regression model for prediction

```python
# importing the dependencies
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

Data pre-processing

```python
# loading the data from csv file to pandas dataframe
salary_data = pd.read_csv("/content/salary_data.csv")
```

```python
# separating the data into features and targets
X = salary_data.iloc[:,:-1].values
Y = salary_data.iloc[:,1].values
```

```python
print(Y)
```

splitting the dataset into training and test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.33, random_state=2)
```

Training the Linear Regression model


```python
model = Linear_Regression(learning_rate = 0.02, no_of_iterations = 1000)
```

```python
model.fit(X_train, Y_train)
```

```python
# printing the model parameters (weights and bias)
print("weight = ", model.w[0])
print("bias = ", model.b)
```

Predict the salary value for test data

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

Visualising the predicted values and actual values

```python
plt.scatter(X_test, Y_test, color="red")
plt.plot(X_test, test_data_prediction, color="blue")
plt.xlabel("work experience")
plt.ylabel("salary")
plt.title("salary vs work experience")
plt.show()
```
